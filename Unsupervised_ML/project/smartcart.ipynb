### Data Preprocessing

### Handle Missing values

```python
df["Income"]=df["Income"].fillna(df["Income"].median())
```

```python
df.head()
```

```python
df.isnull().sum()
```

### Feature Engineering 

```python
df.head()
```

```python
#Age
df["Age"]=2026-df["Year_Birth"]
```

```python
df.head()
```

```python
#Tenure

df["Dt_Customer"]=pd.to_datetime(df["Dt_Customer"],dayfirst=True)

reference_date=df["Dt_Customer"].max()

df["customer_tenure_days"]=(reference_date-df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
df.columns
```

```python
#Spending

df['total_spending']=df['MntWines']+df['MntFruits']+df['MntMeatProducts']+df['MntFishProducts']+df['MntSweetProducts']+df['MntGoldProds']
```

```python
df.head()
```

```python
#children

df['total_children']=df['Kidhome']+df['Teenhome']
```

```python
df.head()
```

```python
#Education

df['Education'].value_counts()

#Graduate,undergraduate and postgraduate 

df['Education']=df['Education'].replace(
    {
        "Basic":"Undergraduate",
        "2n Cycle":"Undergraduate",
        "Graduation":"Graduate",
        "Master":"Postgraduate",
        "PhD":"Postgraduate"
    }
)
```

```python
df["Education"].value_counts()
```

```python
#Marital Status
df.head()
```

```python
df["Marital_Status"].value_counts()
```

```python
df["living_with"]=df["Marital_Status"].replace(
    {
        "Married":"Partner",
        "Together":"Partner",
        "Single":"Alone",
        "Widow":"Alone",
        "YOLO":"Alone",
        "Absurd":"Alone",
        "Divorced":"Alone"
    }
)
```

```python
df["living_with"].value_counts()
```

### Drop Columns

```python
df.head()
```

```python
df.columns
```

```python
cols=["ID","Year_Birth","Marital_Status","Kidhome","Teenhome","Dt_Customer"]
spending_cols=['MntWines', 'MntFruits',
       'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts',
       'MntGoldProds',]

cols_to_drop=cols+spending_cols
```

```python
df_clean=df.drop(columns=cols_to_drop)
```

```python
df_clean.shape
```

```python
df.shape
```

```python
df_clean.head()
```

### Outliers Detection

```python
cols=["Income","Recency","Response","Age","total_spending","total_children"]
```

```python
#relative plots of some features
sns.pairplot(df_clean[cols])
```

```python
df_clean.columns
```

```python
#Remove Outliers

print("data size with outliers",len(df_clean))

df_clean=df_clean[(df_clean["Age"]<90)]
df_clean=df_clean[(df_clean["Income"]<600_000)]
print("data size without outliers",len(df_clean))
```

### Correlation Heatmap

### Feature Encoding

```python
from sklearn.preprocessing import OneHotEncoder 
```

```python
ohe=OneHotEncoder()

cat_cols=["Education","living_with"]

enc_cols=ohe.fit_transform(df_clean[cat_cols])
```

```python
enc_df=pd.DataFrame(enc_cols.toarray(),columns=ohe.get_feature_names_out(cat_cols),index=df_clean.index)
```

```python
enc_df
```

```python
df_encoded=pd.concat([df_clean.drop(columns=cat_cols),enc_df],axis=1)
```

```python
df_encoded.shape
```

```python
df_encoded.head()
```

### Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
X=df_encoded
```

```python
scaler=StandardScaler()

X_scaled=scaler.fit_transform(X)
```

### Visualize

```python
X_scaled.shape
```

```python
#2D visualization

from sklearn.decomposition import PCA

pca=PCA(n_components=2)

X_pca=pca.fit_transform(X_scaled)
```

```python
plt.scatter(X_pca[:,0],X_pca[:,1])
```

```python
pca.explained_variance_ratio_
```

```python
#3d 
pca=PCA(n_components=3)
X_pca=pca.fit_transform(X_scaled)

fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2])

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA")
ax.set_zlabel("PCA3")
```

```python
pca.explained_variance_ratio_
```
