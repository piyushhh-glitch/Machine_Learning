### Data Preprocessing

### Handle Missing values

```python
df["Income"]=df["Income"].fillna(df["Income"].median())
```

### Feature Engineering 

```python
df.head()
```

```python
#Age
df["Age"]=2026-df["Year_Birth"]
```

```python
df.head()
```

```python
#Tenure

df["Dt_Customer"]=pd.to_datetime(df["Dt_Customer"],dayfirst=True)

reference_date=df["Dt_Customer"].max()

df["customer_tenure_days"]=(reference_date-df["Dt_Customer"]).dt.days
```

```python
df.head()
```

```python
df.columns
```

```python
#Spending

df['total_spending']=df['MntWines']+df['MntFruits']+df['MntMeatProducts']+df['MntFishProducts']+df['MntSweetProducts']+df['MntGoldProds']
```

```python
df.head()
```

```python
#children

df['total_children']=df['Kidhome']+df['Teenhome']
```

```python
df.head()
```

```python
#Education

df['Education'].value_counts()

#Graduate,undergraduate and postgraduate 

df['Education']=df['Education'].replace(
    {
        "Basic":"Undergraduate",
        "2n Cycle":"Undergraduate",
        "Graduation":"Graduate",
        "Master":"Postgraduate",
        "PhD":"Postgraduate"
    }
)
```

```python
df["Education"].value_counts()
```

```python
#Marital Status
df.head()
```

```python
df["Marital_Status"].value_counts()
```

```python
df["living_with"]=df["Marital_Status"].replace(
    {
        "Married":"Partner",
        "Together":"Partner",
        "Single":"Alone",
        "Widow":"Alone",
        "YOLO":"Alone",
        "Absurd":"Alone",
        "Divorced":"Alone"
    }
)
```

```python
df["living_with"].value_counts()
```

### Feature Encoding

```python
from sklearn.preprocessing import OneHotEncoder 
```

```python
ohe=OneHotEncoder()

cat_cols=["Education","living_with"]

enc_cols=ohe.fit_transform(df_clean[cat_cols])
```

```python
enc_df=pd.DataFrame(enc_cols.toarray(),columns=ohe.get_feature_names_out(cat_cols),index=df_clean.index)
```

```python
df_encoded=pd.concat([df_clean.drop(columns=cat_cols),enc_df],axis=1)
```

### Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
X=df_encoded
```

```python
scaler=StandardScaler()

X_scaled=scaler.fit_transform(X)
```

### Visualize

```python
X_scaled.shape
```

```python
#2D visualization

from sklearn.decomposition import PCA

pca=PCA(n_components=2)

X_pca=pca.fit_transform(X_scaled)
```

```python
plt.scatter(X_pca[:,0],X_pca[:,1])
```

```python
pca.explained_variance_ratio_
```

```python
#3d 
pca=PCA(n_components=3)
X_pca=pca.fit_transform(X_scaled)

fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2])

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA")
ax.set_zlabel("PCA3")
```

```python
pca.explained_variance_ratio_
```

### Analyze K value

### 1.Elbow method

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss=[]

for k in range(1,11):
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
knee=KneeLocator(range(1,11),wcss,curve="convex",direction="decreasing")
```

```python
optimal_k=knee.elbow
print("best k value:",optimal_k)
```

```python
#plot


plt.plot(range(1,11),wcss,marker="o")

plt.xlabel("K values")
plt.ylabel("wcss value")
```

### 2.Silhouette Score

```python
from sklearn.metrics import silhouette_score

score=[]

for k in range(2,11):
    kmeans=KMeans(n_clusters=k,random_state=42)
    labels=kmeans.fit_predict(X_pca)
    scores=silhouette_score(X_pca,labels)
    score.append(scores)
    #plot


plt.plot(range(2,11),score,marker="o")

plt.xlabel("K values")
plt.ylabel("Silhouette score")
```

### Hence K=4

```python
# combined plot

k_range = range(2, 11)

fig, ax1 = plt.subplots(figsize=(8, 6))

ax1.plot(k_range, wcss[:len(k_range)],
         marker="o",
         color="blue")

ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()

ax2.plot(k_range, score[:len(k_range)],
         marker="x",
         color="red",
         linestyle="--")

ax2.set_ylabel("Silhouette Score")

plt.title("Elbow Method and Silhouette Score")
plt.show("SS")
```

 ### Around 4 it intersects

### Clustering 

```python
#Kmeans

kmeans=KMeans(n_clusters=4,random_state=42)
labels=kmeans.fit_predict(X_pca)
#3d 
pca=PCA(n_components=3)
X_pca=pca.fit_transform(X_scaled)

fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels)
```

```python
#Agglomerative Clustering

from sklearn.cluster import AgglomerativeClustering
```

```python
agglo=AgglomerativeClustering(n_clusters=4,linkage="ward")
labels_agg=agglo.fit_predict(X_pca)
fig=plt.figure(figsize=(8,6))
ax=fig.add_subplot(111,projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2],c=labels_agg)
```

### Hence Agglomerative clustering is better
