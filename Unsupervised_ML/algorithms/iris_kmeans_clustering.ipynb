```python
from sklearn.datasets import load_iris
import seaborn as sns
from sklearn.cluster import KMeans
import pandas as pd
```

```python
iris=load_iris()
```

```python
X=iris.data
y=iris.target
```

```python
#Visualize

sns.scatterplot(x=X[:,0],y=X[:,1],c=y)
```

```python
sns.scatterplot(x=X[:,1],y=X[:,2],c=y)
```

```python
sns.scatterplot(x=X[:,2],y=X[:,3],c=y)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
X_scaled=scaler.fit_transform(X)
```

```python
#PCA -Dimensionality reduction

from sklearn.decomposition import PCA

pca=PCA(
    n_components=2
)
pca_data=pca.fit_transform(X_scaled)
```

```python
#Elbow method

wcss=[]

for k in range(1,21):
    kmeans=KMeans(n_clusters=k)
    
    kmeans.fit_predict(pca_data)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1,21),y=wcss,marker="o")
```

### K=3

```python
kmeans=KMeans(n_clusters=3)
labels=kmeans.fit_predict(pca_data)

sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],c=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,1],marker="o",size=200)
```
