A Web scraping project for Practice

```python
import requests
from bs4 import BeautifulSoup
```

```python
URL="https://books.toscrape.com/"
res=requests.get(URL)
print(res.status_code)
```

```python
res.encoding="utf-8"
```

```python
soup=BeautifulSoup(res.text,"lxml")
books=soup.select("article.product_pod")[:10]

print(len(books))
```

```python
with open("scrap_data/books.html","w") as f:
    f.write(res.text)
```

```python
with open("scrap_data/books.html","r") as f:
    html_content=f.read()

soup=BeautifulSoup(html_content,"lxml")
```

```python
final=[]
for book in books:
    title=book.select_one("h3 a")["title"]
    print(title)
    price=book.select_one("p.price_color").get_text()
    print(price)
    stock=book.select_one("p.instock.availability").get_text().strip()
    print(stock)
    rating=book.select_one("p.star-rating")["class"][1]
    print(rating)
    final.append([title,price,stock,rating])
```

```python
final
```

```python
import pandas as pd
df=pd.DataFrame(final,columns=["Title","Price","Avaibility","Rating"])
```
