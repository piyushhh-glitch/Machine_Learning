```python
import requests
from bs4 import BeautifulSoup
import pandas as pd 
```

```python
URL="https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops"
res=requests.get(URL)
res.status_code
```

```python
soup=BeautifulSoup(res.text,"lxml")
all_laptops=soup.select("div.product-wrapper.card-body")[:20]

with open("scrap_data/laptops","w") as f:
    f.write(res.text)
```

```python
final_list=[]
for laptop in all_laptops:
    name=laptop.select_one("h4 a")["title"]
    print(name)
    price=laptop.select_one("span").get_text()
    print(price)
    description=laptop.select_one("p.description.card-text").get_text()
    print(description)
    rating=laptop.select_one("p[data-rating]")["data-rating"]
    print(rating)
    review=laptop.select_one("p.review-count").get_text().strip()
    print(review)
    final_list.append([name,price,description,rating,review])
```

```python
df=pd.DataFrame(final_list,columns=["name","price","description","rating","review"])
df
```
