```python
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips=sns.load_dataset("tips")
```

```python
print(type(tips))
```

```python
print(tips.head())
```

```python
sns.set_theme(style="white")
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
sns.set_theme()
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="line",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
flights_pivot=flights.pivot(index="month",columns="year",values="passengers")
```

```python
sns.heatmap(
    flights_pivot,annot=True,fmt="d",cmap="coolwarm"
    )
plt.tight_layout()
```

```python
fig,ax=plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    marker="o",
    hue='sex',
    ax=ax,
    ci=None
)
ax.set_title("avg bill value")
ax.set_xlabel("days")
ax.set_ylabel("total bill")
fig.tight_layout()
```
