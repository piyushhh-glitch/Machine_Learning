```python
print(type(tips))
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

```python
sns.set_theme(style="ticks")
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

```python
sns.set_theme(style="dark")
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    
)
```

```python
from matplotlib.pyplot import tight_layout


sns.set_theme()
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker",
    
)
plt.tight_layout()
```
