```python
import requests
from bs4 import BeautifulSoup
```

```python
URL="https://sandbox.oxylabs.io/products"
res=requests.get(URL)
print(res.status_code)
```

```python
soup=BeautifulSoup(res.text,"lxml")
print(soup.title)
```

```python
items=soup.select("div.product-card")
items
```

```python
with open("data/items.html","w") as f:
    f.write(res.text)
```

```python
e_games=[]
```

```python
for item in items:

    title=item.select_one("h4.title").get_text()
    print(title)

    price=item.select_one("div.price-wrapper").get_text()
    print(price)

    categ=item.select_one("p.category")
    spans=categ.find_all("span")
    category=[]
    for span in spans:
        category.append(span.get_text(strip=True))
    print(category)

    description=item.select_one("p.description").get_text()
    print(description)

    link=item.select_one("a.card-header")["href"]
    full_link="https://sandbox.oxylabs.io"+link
    print(full_link)

    e_games.append([title,price,category,description,full_link])
```

```python
e_games
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(e_games,columns=["Title","Price","Category","Description","Links"])
```

```python
df
```

```python
df.to_csv("data/games.csv",index=False)
```
