```python
df=pd.read_sql("SELECT * FROM items",conn)
conn.close()
```

```python
# 1.Data Loading
print(df.head())
print(df.tail())
print(df.info())
print(df.describe())
```

```python
# 2.checking duplicate values and null values
print(df.duplicated())
print(df.isnull())
```

```python
# 3.price cleaning
print(df.dtypes)
print(df["price"].isnull())
print(df["price"].min())
print(df["price"].max())
```

```python
# 4.Feature engineering

def price_range(p):
    if p<70:
        return "cheap"
    elif p<85:
        return "mid"
    else:
        return "expensive"
# 1.Price range
df["price_range"]=df["price"].apply(price_range)
#2.Lenght of characters in description
df["desc_lenght"]=df["description"].apply(len)
#3.Number of categories
df["categ_count"]=df["category"].apply(lambda x:len(str(x).split(",")))
df
```

```python
#5.Data extraction
#1.Expensive
top_expensive=df.sort_values("price",ascending=False).head(5)
print(top_expensive)
#2.Cheapest
cheapest=df.sort_values("price").tail()
print(cheapest)

#3.Simple Filter
filtered=df[df["price"]>70]
print(filtered.head())

```

```python
#6.Groupby

df["category"] = df["category"].str.replace("[\[\]']", "", regex=True)
df["category"]=df["category"].str.split(",")
df=df.explode("category")
df.groupby("category").size().sort_values(ascending=False)
```
