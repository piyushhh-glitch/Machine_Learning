```python
import requests
from bs4 import BeautifulSoup
```

```python
URL="https://realpython.github.io/fake-jobs/"
res=requests.get(URL)
print(res.status_code)
```

```python
soup=BeautifulSoup(res.text,"lxml")
jobs=soup.select("div.card-content")[:30]
print(len(jobs))
```

```python
with open("scrap_data/jobs.html","w") as f:
    f.write(res.text)
```

```python
jobs_list=[]
```

```python
for job in jobs:
    title=job.find("h2",class_="title is-5").get_text()
    print(title)
    company=job.find("h3",class_="subtitle is-6 company").get_text()
    print(company)
    location=job.select_one("p.location").get_text().strip()
    print(location)
    apply=job.select_one("a.card-footer-item").get("href")
    print(apply)
    jobs_list.append([title,company,location,apply])
```

```python
import pandas as pd 
```

```python
df=pd.DataFrame(jobs_list,columns=["Title","Company","Location","Apply Link"])
```
